```python
import os
```

```python
# 1. Define and create the folder structure
folders = ["notebooks", "sql", "docs"]
for folder in folders:
    os.makedirs(folder, exist_ok=True)
    print(f"Created directory: {folder}/")
```

```python
# 2. Write the .gitignore file
gitignore_content = """# Databricks and Jupyter temporary files
.ipynb_checkpoints/
.virtual_documents/
*.pyc

# Local IDEs (VS Code / PyCharm)
.vscode/
.idea/
*.log

# OS specific files
.DS_Store
Thumbs.db

# Environmental variables / credentials
.env
*.secret
"""

with open(".gitignore", "w") as f:
    f.write(gitignore_content)
print("Created .gitignore")
```
